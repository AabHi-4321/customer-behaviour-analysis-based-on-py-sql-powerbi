```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()) )
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# create a columns age group
df['age_group']=pd.qcut(df['age'], q=4, labels=['Young Adult','Adult','Middle-aged','Senior'])
```

```python
df[['age','age_group']].head()
```

```python
# create column purchase_frequency_days
unique_values = df['frequency_of_purchases'].unique()
# print(unique_values)

frequency_mapping={
   'Fortnightly':14,
    'Weekly':7,
    'Annually' :365,
    'Quarterly' :15,
    'Bi-Weekly':14,
    'Monthly':30,
 'Every 3 Months':90 
}

df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
df[['purchase_frequency_days','frequency_of_purchases']].head()
```
